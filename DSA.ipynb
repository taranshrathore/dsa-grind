```python
#alright two sum leetcode using nested loop

def two_sum(nums, target):
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                return [i, j]

nums = [2, 7, 11, 15]
target = 9

print(two_sum(nums, target))  # [0, 1]
```

```python
#optimal appraoch using hash map

def two_sum(nums, target):
    seen = {}

    for i in range(len(nums)):
        needed = target - nums[i]

        if needed in seen:
            return [seen[needed], i]

        seen[nums[i]] = i

    return []
```

```python
#Contains duplicate question - Q - if any number in the array appears twice return True else False - set

def contains_duplicate(nums):

    seen = set()

    for num in nums:
        if num in seen:
            return True
    seen.add(num)

    return False

            
```

```python
class Solution:
    def uniqueOccurrences(self, arr: list[int]) -> bool:
        counts = {}

        for num in arr:
            counts[num] = counts.get(num, 0) + 1

        frequencies = set()

        for count in counts.values():
            if count in frequencies:
                return False
            frequencies.add(count)

        return True
```
