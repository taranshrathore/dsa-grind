````python
def is_armstrong(num):
    digits = [int(d) for d in str(num)]
    power = len(digits)
    return num == sum(d ** power for d in digits)
````

````python
def find_middle_sum(arr1, arr2):
    merged = sorted(arr1 + arr2)
    n = len(merged)
    mid1 = merged[n // 2 - 1]
    mid2 = merged[n // 2]
    return mid1 + mid2

n = int(input())
arr1 = list(map(int, input().split()))
arr2 = list(map(int, input().split()))
print(find_middle_sum(arr1, arr2))
````

````python
def count_correct_positions(heights):
    expected = sorted(heights)
    count = 0
    for actual, expected_height in zip(heights, expected):
        if actual == expected_height:
            count += 1
    return count

heights = [1, 1, 4, 2, 1, 3]
print(count_correct_positions(heights)) 
````

````python
def is_transformed_palindrome(S):
    n = len(S)
    mid = n // 2

    if n % 2 == 0:
        left = S[:mid]
        right = S[mid:]
    else:
        left = S[:mid]
        right = S[mid:]  

    T = left[::-1] + right[::-1]
    return T == T[::-1]

S = input("Enter any string:")
print(is_transformed_palindrome(S))
````

````python
def compress_string(s, n):
    result = []
    i = 0
    while i < len(s):
        count = 1
        while i + count < len(s) and s[i] == s[i + count]:
            count += 1
        if count >= n:
            result.append(f"{s[i]}{count}")
        else:
            result.append(s[i] * count)
        i += count
    return ''.join(result)

s = input("Enter the string: ")
n = int(input("Enter the minimum count for compression: "))
compressed = compress_string(s, n)

````

````python
def max_meetings(meetings):
    meetings.sort(key=lambda x: x[1])
    
    count = 0
    last_end_time = 0
    
    for start, end in meetings:
        if start >= last_end_time:
            count += 1
            last_end_time = end
            
    return count
meetings = [(1, 3), (2, 4), (3, 5), (0, 6), (5, 7), (8, 9), (5, 9)]
print("Maximum number of meetings that can be scheduled:", max_meetings(meetings))
````

````python
import sys
input = sys.stdin.readline

def solve():
    MOD = 10**9 + 7
    
    n, q = map(int, input().split())
    A = list(map(int, input().split()))
    
    total = 0
    
    for _ in range(q):
        query = list(map(int, input().split()))
        t, l, r = query[0], query[1], query[2]
        
        # Convert to 0-indexed
        l -= 1
        r -= 1
        
        if t == 1:
            # Replace A[i] with (i - l + 1) * A[i] for l <= i <= r
            for i in range(l, r + 1):
                A[i] = (A[i] * (i - l + 1)) % MOD
        else:
            # Sum from l to r
            total = (total + sum(A[l:r + 1])) % MOD
    
    print(total)

solve()
```

### Example Trace
```
A = [1, 2, 3, 4, 5]
Type 1, l=2, r=4 (1-indexed):
  i=2: A[2] = (2-2+1)*2 = 1*2 = 2
  i=3: A[3] = (3-2+1)*3 = 2*3 = 6
  i=4: A[4] = (4-2+1)*4 = 3*4 = 12
  A = [1, 2, 6, 12, 5]

Type 2, l=1, r=5 → sum = 1+2+6+12+5 = 26
````

````python
def dfs(graph, start, visited=None):
    if visited is None:
        visited = set()
    visited.add(start)
    print(start)  # Process the node (for example, print it)
    
    for neighbor in graph[start]:
        if neighbor not in visited:
            dfs(graph, neighbor, visited)
    
    return visited

# Example usage:
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

dfs(graph, 'A')
````

````python
class TreeNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

def insert(root, value):
    if root is None:
        return TreeNode(value)
    else:
        if value < root.value:
            root.left = insert(root.left, value)
        else:
            root.right = insert(root.right, value)
    return root

def pre_order_traversal(root):
    if root:
        print(root.value, end=' ')
        pre_order_traversal(root.left)
        pre_order_traversal(root.right)

````

````python
class Solution:
    def findCircleNum(self, isConnected: List[List[int]]) -> int:
        V = len(isConnected)
        visited = [False] * V

        def dfs(u):
            for v in range(V):
                if isConnected[u][v] == 1 and not visited[v]:
                    visited[v] = True
                    dfs(v)

        provinces = 0
        for i in range(V):
            if not visited[i]:
                visited[i] = True
                dfs(i)
                provinces += 1

        return provinces
````

````python
def max_sum_good_subarray(A, K):
    n = len(A)
    left = 0
    right = 0
    max_sum = 0
    current_sum = 0
    freq_map = {}

    while right < n:
        if A[right] in freq_map:
            freq_map[A[right]] += 1
        else:
            freq_map[A[right]] = 1
        
        current_sum += A[right]
        
        while len(freq_map) > K:
            freq_map[A[left]] -= 1
            current_sum -= A[left]
            if freq_map[A[left]] == 0:
                del freq_map[A[left]]
            left += 1
        
        max_sum = max(max_sum, current_sum)
        
        right += 1

    return max_sum

A = [1, 2, 1, 2, 3]
K = 2
print("Maximum sum of a good subarray:", max_sum_good_subarray(A, K))
````

````python
class SegmentTree:
    def __init__(self, data):
        self.n = len(data)
        self.tree = [0] * (2 * self.n)
        self.build(data)

    def build(self, data):
        for i in range(self.n):
            self.tree[self.n + i] = data[i]
        for i in range(self.n - 1, 0, -1):
            self.tree[i] = self.tree[i * 2] + self.tree[i * 2 + 1]

    def update(self, index, value):
        index += self.n
        self.tree[index] = value
        while index > 1:
            index //= 2
            self.tree[index] = self.tree[index * 2] + self.tree[index * 2 + 1]

    def query(self, left, right):
        result = 0
        left += self.n
        right += self.n
        while left < right:
            if left & 1:
                result += self.tree[left]
                left += 1
            if right & 1:
                right -= 1
                result += self.tree[right]
            left //= 2
            right //= 2
        return result

data = [1, 3, 5, 7, 9, 11]
seg_tree = SegmentTree(data)
print("Sum of values in range [1, 4):", seg_tree.query(1, 4))
seg_tree.update(1, 10)
print("Sum of values in range [1, 4):", seg_tree.query(1, 4))
````
